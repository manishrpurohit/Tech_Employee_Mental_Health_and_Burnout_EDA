## Setup and Data Loading


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import scipy.stats as stats

# Load the dataset
df = pd.read_csv(r"D:\Machine_learning\DataSet\mental_health_burnout_tech_2026.csv", header=0, index_col=0)
```

### 1. Histogram plot
**Definition:** A chart that groups continuous data into ranges called 'bins' and uses bars to show frequency.

**Purpose:** To quickly see the distribution and spread of a continuous variable.


```python
df['age'].plot(kind='hist')
plt.show()
```

**Inference:** Shows the most common age range for tech employees in the dataset.


### 2. Density plot
**Definition:** A smoothed curve representing the distribution of numeric data.

**Purpose:** To identify peaks in the data's probability distribution without jagged bins.


```python
df['burnout_score'].plot(kind='density')
plt.show()
```

**Inference:** Shows the smooth curve of burnout scores, highlighting the most frequent score.


### 3. KDE plot
**Definition:** Kernel Density Estimate overlays multiple smooth density distributions.

**Purpose:** To compare the distributions of different categories side-by-side.


```python
sns.kdeplot(data=df, x='stress_score', hue='gender')
plt.show()
```

**Inference:** Shows if different genders have different peaks in stress levels.


### 4. Pie chart
**Definition:** A circular chart divided into slices based on proportions.

**Purpose:** To show the relative percentage of categories compared to the whole.


```python
df['work_mode'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.ylabel('')
plt.show()
```

**Inference:** Shows the proportion of Remote, Hybrid, and On-site workers.


### 5. Donut chart
**Definition:** A pie chart with a blank center.

**Purpose:** To display part-to-whole relationships cleanly.


```python
plt.pie(df['seniority_level'].value_counts(), labels=df['seniority_level'].value_counts().index, wedgeprops=dict(width=0.3))
plt.show()
```

**Inference:** Shows the ratio of different seniority levels as a ring.


### 6. Line chart
**Definition:** Data points connected by straight line segments.

**Purpose:** To show trends over a continuous sequence.


```python
df.groupby('years_experience')['burnout_score'].mean().plot(kind='line')
plt.ylabel('Average Burnout')
plt.show()
```

**Inference:** Shows how the average burnout score changes as years of experience increase.


### 7. Multi-line chart
**Definition:** Multiple line charts plotted on the same axes.

**Purpose:** To compare the trends of different sub-groups.


```python
df.pivot_table(index='years_experience', columns='gender', values='burnout_score').plot(kind='line')
plt.ylabel('Average Burnout')
plt.show()
```

**Inference:** Compares how burnout progresses over experience across different genders.


### 8. Bar plot
**Definition:** Rectangular bars representing categorical values.

**Purpose:** To compare numerical values across distinct categories.


```python
sns.barplot(data=df, x='company_size', y='burnout_score')
plt.show()
```

**Inference:** Cleanly shows which company sizes (Small, Medium, Large) have the highest and lowest average burnout scores.


### 9. Dual-axis line chart
**Definition:** A line chart with two different Y-axes.

**Purpose:** To compare two variables with completely different scales.


```python
ax = df.groupby('years_experience')['stress_score'].mean().plot(color='green', label='Stress', legend=True)
df.groupby('years_experience')['phq9_score'].mean().plot(secondary_y=True, ax=ax, color='blue', label='PHQ-9', legend=True)
plt.show()
```

**Inference:** Compares average stress and depression (PHQ-9) over years of experience.


### 10. Stacked bar plot
**Definition:** A bar chart where each bar is divided into colored segments.

**Purpose:** To show total size and internal composition simultaneously.


```python
pd.crosstab(df['seniority_level'], df['burnout_level']).plot(kind='bar', stacked=True)
plt.show()
```

**Inference:** Shows total employees per seniority, broken down by burnout level.


### 11. Clustered barplot
**Definition:** Bars for sub-categories placed side-by-side in groups.

**Purpose:** To easily compare multiple sub-groups directly across primary categories.


```python
sns.barplot(data=df, x='work_mode', y='burnout_score', hue='gender')
plt.show()
```

**Inference:** Compares burnout by work mode, split by gender.


### 12. Heat map
**Definition:** A matrix where data values are represented by color gradients.

**Purpose:** To identify clusters and highs/lows in cross-tabulated data.


```python
sns.heatmap(df.pivot_table(index='work_mode', columns='seniority_level', values='burnout_score'), annot=True)
plt.show()
```

**Inference:** The grid of colors helps us instantly spot which combinations of Work Mode and Seniority have the worst average burnout scores.


### 13. Simple Correlation chart
**Definition:** A bar chart showing correlation values for a target variable.

**Purpose:** To quickly see which numerical features are most positively or negatively correlated with a target.


```python
df.select_dtypes('number').corr()['burnout_score'].drop('burnout_score').plot(kind='bar')
plt.ylabel('Correlation with Burnout Score')
plt.show()
```

**Inference:** Shows which numeric variables increase or decrease alongside burnout score.


### 14. Scatter plot
**Definition:** Individual points plotted on an X and Y axis.

**Purpose:** To visually inspect the relationship between two continuous variables.


```python
df.sample(300, random_state=42).plot.scatter(x='autonomy_score', y='stress_score', alpha=0.6, color='purple')
plt.show()
```

**Inference:** By plotting two continuous decimal variables (Autonomy vs Stress), we get a nicely spread out cloud of points. We can observe if having higher autonomy scatters towards lower stress.


### 15. Bubble chart
**Definition:** A scatter plot where points have varying sizes.

**Purpose:** To compare three numeric variables simultaneously (X, Y, and size).


```python
agg_df = df.groupby('job_role').agg(stress=('stress_score', 'mean'), burnout=('burnout_score', 'mean'), count=('age', 'count')).reset_index()
sns.scatterplot(data=agg_df, x='stress', y='burnout', size='count', sizes=(50, 500))
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

**Inference:** Each bubble is a Job Role. We can see average stress vs average burnout, while the bubble size perfectly shows how many people work in that role.


### 16. Lollipop chart
**Definition:** A line ending in a dot, acting as a minimalist bar chart.

**Purpose:** To present categorical comparisons cleanly with less clutter.


```python
industry_burnout = df.groupby('industry')['burnout_score'].mean()
plt.stem(industry_burnout.index, industry_burnout.values)
plt.xticks(rotation=90)
plt.show()
```

**Inference:** Cleanly displays the average burnout score per industry.


### 17. Tree map
**Definition:** Nested rectangles used to display hierarchical data.

**Purpose:** To show the proportion of categories. Larger rectangles mean larger quantities.


```python
fig = px.treemap(df.dropna(subset=['industry', 'job_role']), path=['industry', 'job_role'])
fig.show()
```

**Inference:** Shows the hierarchical breakdown of the workforce from Industry down to Job Role.


### 18. Sunburst diagram
**Definition:** A multilevel pie chart consisting of inner and outer rings.

**Purpose:** To display hierarchical data radiating outwards from the center.


```python
fig = px.sunburst(df.dropna(subset=['work_mode', 'seniority_level']), path=['work_mode', 'seniority_level'])
fig.show()
```

**Inference:** Visually breaks down the company by Work Mode, then by Seniority Level in circular rings.


### 19. Box plot
**Definition:** A plot summarizing data using the median, quartiles, and outliers.

**Purpose:** To quickly display spread and explicitly highlight outliers.


```python
sns.boxplot(data=df, x='work_mode', y='stress_score')
plt.show()
```

**Inference:** Shows the median stress and highlights any extreme outlier stress scores per work mode.


### 20. Whisker plot
**Definition:** A simple box-and-whisker plot focusing on the lines (whiskers) extending from quartiles.

**Purpose:** To visually indicate the range of data outside the middle 50%.


```python
df['burnout_score'].dropna().plot(kind='box')
plt.show()
```

**Inference:** The simple box and whiskers show the total range of all burnout scores in the company.


### 21. Parallel plot
**Definition:** A chart plotting multivariate data on vertical parallel axes.

**Purpose:** To compare many numerical variables to find clusters or patterns.


```python
from pandas.plotting import parallel_coordinates
subset = df.groupby('seniority_level')[['autonomy_score', 'stress_score', 'burnout_score']].mean().reset_index()
parallel_coordinates(subset, 'seniority_level')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

**Inference:** Plots a single clean line for each seniority level, showing how their average health metrics behave across different scoring categories.


### 22. Swarm plot
**Definition:** A scatter plot where points are adjusted to not overlap.

**Purpose:** To see the exact granular distribution of points for small datasets.


```python
sns.swarmplot(data=df.sample(200, random_state=42), x='burnout_level', y='stress_score')
plt.show()
```

**Inference:** Physically shows the count and shape of stress scores within burnout levels.


### 23. Area chart
**Definition:** A line chart with the space below filled with color.

**Purpose:** To emphasize the sheer volume of a metric over a continuous axis.


```python
df.groupby('years_experience')['stress_score'].sum().plot(kind='area')
plt.ylabel('Total Stress')
plt.show()
```

**Inference:** Emphasizes the total accumulated stress across different years of experience.


### 24. Stacked area chart
**Definition:** Multiple area charts stacked on top of each other.

**Purpose:** To show how sub-parts contribute to a total volume over time.


```python
pd.crosstab(df['years_experience'], df['burnout_level']).plot(kind='area')
plt.show()
```

**Inference:** Shows the volume of each burnout level contributing to the total workforce at each experience year.


### 25. Violin plot
**Definition:** A combination of a box plot and a density plot.

**Purpose:** To visualize both summary statistics and exact density distributions.


```python
sns.violinplot(data=df, x='burnout_level', y='phq9_score')
plt.show()
```

**Inference:** Shows where the majority of depression scores lie for each burnout tier.


### 26. Combo chart
**Definition:** A chart combining two visual types, like a bar and a line chart.

**Purpose:** To simultaneously display two different metrics on dual axes.


```python
ax = df.groupby('job_role')['age'].count().plot(kind='bar', color='lightblue')
df.groupby('job_role')['burnout_score'].mean().plot(kind='line', secondary_y=True, ax=ax, color='red', marker='o')
plt.show()
```

**Inference:** Compares the number of employees (bars) to their average burnout (red line) per job role.


### 27. Pareto chart
**Definition:** A sorted bar chart combined with a cumulative percentage line.

**Purpose:** To identify the top categories making up the vast majority of the total.


```python
counts = df['country'].value_counts()
ax = counts.plot(kind='bar')
(counts.cumsum() / counts.sum() * 100).plot(kind='line', secondary_y=True, ax=ax, color='orange', marker='x')
plt.show()
```

**Inference:** The orange line helps identify which countries make up 80% of the dataset.


### 28. Slope chart
**Definition:** A line chart drawn between exactly two points.

**Purpose:** To clearly emphasize the change between a 'Before' and 'After' state.


```python
df.groupby('seniority_level')['burnout_score'].mean()[['Junior', 'Senior']].plot(kind='line', marker='o')
plt.ylabel('Avg Burnout')
plt.show()
```

**Inference:** Shows the direct shift in average burnout when moving from Junior to Senior level.


### 29. Column chart
**Definition:** A standard vertical bar chart.

**Purpose:** To compare categorical values vertically.


```python
df.groupby('industry')['seeks_mental_health_support'].sum().plot(kind='bar')
plt.ylabel('Total Seeking Support')
plt.show()
```

**Inference:** Simply compares total employees seeking mental health support by industry.


### 30. Q-Q plot
**Definition:** A plot mapping data quantiles against a theoretical bell curve.

**Purpose:** To visually verify whether a dataset follows a normal distribution.


```python
stats.probplot(df['burnout_score'].dropna(), dist='norm', plot=plt)
plt.show()
```

**Inference:** If the dots stray from the red line, the burnout scores are not perfectly normally distributed.


### 31. Strip plot
**Definition:** A 1D scatter plot where all dots for a category are on a straight line.

**Purpose:** To see the absolute max and min points.


```python
sns.stripplot(data=df.sample(200, random_state=42), x='burnout_level', y='stress_score', jitter=False)
plt.show()
```

**Inference:** Without jitter, it draws straight lines showing the absolute range of stress scores.


### 32. Jitter plot
**Definition:** A Strip plot with randomized horizontal noise.

**Purpose:** To spread out overlapping points to visualize density.


```python
sns.stripplot(data=df.sample(200, random_state=42), x='burnout_level', y='stress_score', jitter=True)
plt.show()
```

**Inference:** Spreads out the dots so you can see where the clusters of stress scores are.


## Definitions of Unplotted Graphs

Below are the definitions and purposes for the remaining graphs that were not plotted above.

### 33. Radar chart
**Definition:** A spider web-like plot with multiple axes radiating from a center point.
**Purpose:** To compare a subject across several different quantitative variables on standardized scales.

### 34. Waterfall chart
**Definition:** A chart illustrating how an initial value is affected by a series of intermediate positive or negative values.
**Purpose:** To understand the cumulative effect of sequentially introduced data points.

### 35. Ridge plot
**Definition:** A series of density plots stacked vertically, partially overlapping each other.
**Purpose:** To compare the shape of distributions across multiple categories beautifully, saving horizontal space.

### 36. Hexbin map
**Definition:** A scatter plot that divides the background into hexagons, coloring them by the density of points inside.
**Purpose:** To handle extremely large datasets where standard scatter plots just turn into an unreadable solid blob.

### 37. Dot plot
**Definition:** A chart that plots data points along a single axis.
**Purpose:** Best used for small datasets to show the distribution and frequency of individual items without grouping them into bins.

### 38. Network chart
**Definition:** A map of dots (nodes) connected by lines (edges).
**Purpose:** Used to visualize relationships, interactions, and structures within complex networks (like social media friends or computer networks).

### 39. Word cloud
**Definition:** A cluster of words where the size of the text corresponds to its frequency.
**Purpose:** Used to quickly identify the most prominent keywords or themes in a large body of text.

### 40. Venn diagram
**Definition:** Overlapping circles.
**Purpose:** Used to show the logical relationships, commonalities, and differences between different distinct groups or sets.

### 41. Dendrogram
**Definition:** A branching, tree-like diagram.
**Purpose:** Used heavily in machine learning to illustrate the arrangement of clusters produced by hierarchical clustering.

### 42. Chord diagram
**Definition:** A circular diagram where entities are arranged around a circle and curves connect them.
**Purpose:** Used to visualize inter-relationships or data flow between entities (e.g., migration between countries).

### 43. Sankey chart
**Definition:** A flow diagram in which the width of the arrows is proportional to the flow rate.
**Purpose:** Excellent for visualizing the flow of money, energy, or materials through different stages of a system.

### 44. Arc diagram
**Definition:** A network graph where all nodes are placed on a single straight horizontal line, connected by semi-circular arcs.
**Purpose:** Used to find hidden structures or sequential connections in a network.

### 45. Gantt chart
**Definition:** A type of horizontal bar chart.
**Purpose:** Widely used in project management to illustrate a project schedule, showing task start and end dates.

### 46. Choropleth map
**Definition:** A geographical map where areas are shaded or patterned.
**Purpose:** Used to represent statistical variables through color mapping on geographic regions (e.g., population density by state).

### 47. Bump chart
**Definition:** A specific type of line chart.
**Purpose:** Used exclusively to track the changes in rank or position of different items over time, ignoring their absolute values.

### 48. Connection map
**Definition:** A geographic map with straight or curved lines connecting points.
**Purpose:** Used to display spatial routes, connections, or flows between different geographic locations (like flight paths).

### 49. Stream graph
**Definition:** A type of stacked area chart displaced around a central axis, resulting in a flowing, organic shape.
**Purpose:** Used to visualize high-volume datasets to discover trends over time with a strong aesthetic appeal.

### 50. Circular bar chart
**Definition:** A bar chart plotted on a polar coordinate system instead of a straight line.
**Purpose:** Used for its visual appeal or to show cyclical patterns (like months of the year), though it can be harder to read precisely.

### 51. Polar area chart
**Definition:** Similar to a pie chart, but all slices have the same angle, and the value is shown by how far the slice extends outwards (the radius).
**Purpose:** Used to display cyclical data where magnitudes vary greatly.

### 52. Multilevel pie/donut chart
**Definition:** Concentric rings of pie/donut charts.
**Purpose:** Just like a sunburst, used to show the breakdown of hierarchical categories.

### 53. Hive plot
**Definition:** A network visualization where nodes are placed on radial axes and connected by curves.
**Purpose:** Designed to create rational, reproducible layouts for large, complex networks to avoid the "hairball" effect.

### 54. Flow plot
**Definition:** A general category of charts mapping movement.
**Purpose:** Used to track how quantities or entities move or transition from one state to another.

### 55. Pictorial chart
**Definition:** A chart that uses icons or pictures instead of bars or lines.
**Purpose:** Highly effective for infographics and engaging non-technical audiences by making data visually literal.

### 56. Pyramid chart
**Definition:** A chart in the shape of a triangle divided into horizontal sections.
**Purpose:** Used to show hierarchical structures where the base is broad and the top is narrow (like population demographics).

### 57. Euler's chart
**Definition:** Similar to a Venn diagram, but the area of the shapes is proportional to the number of items, and sets that don't overlap aren't forced to intersect.
**Purpose:** Used for a more accurate visual representation of complex set relationships.

### 58. Process diagram
**Definition:** A flowchart.
**Purpose:** Used to map out the logical, sequential steps in a process or decision tree.

### 59. Stacked column chart
**Definition:** A vertical version of the stacked bar plot.
**Purpose:** Used to show part-to-whole relationships across categories vertically.

### 60. Grouped column chart
**Definition:** A vertical version of the clustered bar plot.
**Purpose:** Used to directly compare multiple sub-groups within main categories side-by-side.

### 61. Waffle chart
**Definition:** A square grid, usually 10x10, where colored squares represent percentages.
**Purpose:** Excellent for visualizing progress towards a goal or exactly showing a percentage out of 100%.

### 62. Control chart
**Definition:** A line chart with average, upper, and lower control limit lines.
**Purpose:** Essential in manufacturing and quality control to track if a process is stable or behaving unpredictably.

### 63. Curve bar chart
**Definition:** A bar chart where the bars have heavily rounded or curved edges.
**Purpose:** Primarily an aesthetic choice used in modern dashboard design for a softer, sleeker look.

### 64. Drilldown/Drillup chart
**Definition:** An interactive chart that allows clicking to see more detailed data.
**Purpose:** Used in business intelligence software to let users navigate seamlessly from summary data down to granular data.

### 65. Timer
**Definition:** A visual gauge, progress bar, or countdown.
**Purpose:** Used in dashboards to display time elapsed or time remaining against a specific temporal goal.
